```python
# Cloud-Based Online Quiz System
# Python program for Google Colab

import sqlite3

# -------------------------------
# Create Database
# -------------------------------

connection = sqlite3.connect("online_quiz.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS results (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    student_name TEXT,
    score INTEGER,
    total_questions INTEGER
)
""")

connection.commit()


# -------------------------------
# Quiz Questions
# -------------------------------

questions = [
    {
        "question": "What is cloud computing?",
        "options": [
            "A. Storing and accessing resources over the Internet",
            "B. Only using a personal computer",
            "C. A type of keyboard",
            "D. A programming language"
        ],
        "answer": "A"
    },

    {
        "question": "Which of the following is a cloud service model?",
        "options": [
            "A. HTTP",
            "B. SaaS",
            "C. HTML",
            "D. CPU"
        ],
        "answer": "B"
    },

    {
        "question": "What does IaaS stand for?",
        "options": [
            "A. Internet as a Service",
            "B. Infrastructure as a Service",
            "C. Information as a System",
            "D. Internet Application Service"
        ],
        "answer": "B"
    },

    {
        "question": "What does SaaS stand for?",
        "options": [
            "A. Software as a Service",
            "B. Storage as a System",
            "C. Software and Storage Service",
            "D. System as a Service"
        ],
        "answer": "A"
    },

    {
        "question": "Which is a popular cloud platform?",
        "options": [
            "A. Amazon Web Services",
            "B. Microsoft Paint",
            "C. Notepad",
            "D. Calculator"
        ],
        "answer": "A"
    },

    {
        "question": "Which technology allows multiple virtual machines to run on one physical machine?",
        "options": [
            "A. Virtualization",
            "B. Encryption",
            "C. Compilation",
            "D. Formatting"
        ],
        "answer": "A"
    },

    {
        "question": "Which cloud deployment model is available to the general public?",
        "options": [
            "A. Private Cloud",
            "B. Public Cloud",
            "C. Local Cloud",
            "D. Personal Cloud"
        ],
        "answer": "B"
    },

    {
        "question": "Which service model provides a platform for application development?",
        "options": [
            "A. IaaS",
            "B. PaaS",
            "C. SaaS",
            "D. LAN"
        ],
        "answer": "B"
    }
]


# -------------------------------
# Start Quiz
# -------------------------------

print("=" * 50)
print("       CLOUD-BASED ONLINE QUIZ SYSTEM")
print("=" * 50)

student_name = input("Enter your name: ").strip()

while student_name == "":
    print("Name cannot be empty.")
    student_name = input("Enter your name: ").strip()


print("\nWelcome,", student_name)
print("Answer all the questions.")
print("=" * 50)


# -------------------------------
# Conduct Quiz
# -------------------------------

score = 0

for number, question in enumerate(questions, start=1):

    print("\nQuestion", number)
    print(question["question"])

    for option in question["options"]:
        print(option)

    while True:
        answer = input("Enter your answer (A/B/C/D): ").strip().upper()

        if answer in ["A", "B", "C", "D"]:
            break

        print("Please enter only A, B, C, or D.")

    if answer == question["answer"]:
        score += 1


# -------------------------------
# Calculate Result
# -------------------------------

total_questions = len(questions)
percentage = (score / total_questions) * 100


# -------------------------------
# Store Result in Database
# -------------------------------

cursor.execute(
    """
    INSERT INTO results
    (student_name, score, total_questions)
    VALUES (?, ?, ?)
    """,
    (student_name, score, total_questions)
)

connection.commit()


# -------------------------------
# Display Result
# -------------------------------

print("\n" + "=" * 50)
print("              QUIZ RESULT")
print("=" * 50)

print("Student Name :", student_name)
print("Score        :", score, "/", total_questions)
print("Percentage   :", round(percentage, 2), "%")

if percentage >= 80:
    print("Grade        : Excellent")

elif percentage >= 60:
    print("Grade        : Good")

elif percentage >= 40:
    print("Grade        : Pass")

else:
    print("Grade        : Fail")

print("=" * 50)


# -------------------------------
# Display Previous Results
# -------------------------------

print("\nPrevious Quiz Results")
print("-" * 50)

cursor.execute("""
SELECT student_name, score, total_questions
FROM results
ORDER BY id DESC
""")

results = cursor.fetchall()

for result in results:
    name, student_score, total = result
    percentage_result = (student_score / total) * 100

    print(
        "Name:",
        name,
        "| Score:",
        student_score,
        "/",
        total,
        "| Percentage:",
        round(percentage_result, 2),
        "%"
    )


# -------------------------------
# Close Database
# -------------------------------

connection.close()

print("\nQuiz completed successfully.")
```
